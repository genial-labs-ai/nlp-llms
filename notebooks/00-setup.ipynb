## 2. API keys (optional)

Some labs call commercial APIs. **Keys are optional**: with none set, those labs run on a small open model, and you can still complete every exercise.

To add keys in Colab, click the key icon in the left sidebar (**Secrets**), add any of the names below, and switch on **Notebook access**:

| Secret name | Provider |
|---|---|
| `OPENAI_API_KEY` | OpenAI |
| `ANTHROPIC_API_KEY` | Anthropic (Claude) |
| `TYPESAFE_API_KEY` | TypeSafe (Jev) |

Never paste a key into a cell. Outside Colab, set environment variables with the same names.

```python
import os


def get_secret(name: str) -> str | None:
    """Return a secret from Colab Secrets or the environment, or None."""
    try:
        from google.colab import userdata

        return userdata.get(name)
    except Exception:
        return os.environ.get(name)


SECRETS = {
    "openai": "OPENAI_API_KEY",
    "anthropic": "ANTHROPIC_API_KEY",
    "typesafe": "TYPESAFE_API_KEY",
}
KEYS = {provider: get_secret(name) for provider, name in SECRETS.items()}

for provider, name in SECRETS.items():
    status = "found" if KEYS[provider] else "not set"
    print(f"{name:<20} {status}")
```

## 3. Which provider the labs will use

Each API lab has one `PROVIDER` switch. By default it picks the first provider you have a key for, and falls back to an open model when you have none. You can always set it by hand.

```python
PROVIDER = next((p for p in ("openai", "anthropic") if KEYS[p]), "open")
DECISION_MODEL = "jev" if KEYS["typesafe"] else "toy"

print(f"Text generation: {PROVIDER}")
print(f"Decision model:  {DECISION_MODEL}")
if PROVIDER == "open":
    print("No API keys found. The labs will use open models. That is fine.")
```
